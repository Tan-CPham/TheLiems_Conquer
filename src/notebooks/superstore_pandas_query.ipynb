# **1. Load data**

```python
import pandas as pd
```

```python
data = pd.read_csv('/content/superstore.csv')
df = pd.DataFrame(data)
print("rows,cols:", df.shape)
```

```python
new_column_names = {
    'Row ID': 'row_id',
    'Order ID': 'order_id',
    'Order Date': 'order_date',
    'Ship Date': 'ship_date',
    'Ship Mode': 'ship_mode',
    'Customer ID': 'customer_id',
    'Customer Name': 'customer_name',
    'Segment': 'segment',
    'Country': 'country',
    'City': 'city',
    'State': 'state',
    'Postal Code': 'postal_code',
    'Region': 'region',
    'Product ID': 'product_id',
    'Category': 'category',
    'Sub-Category': 'sub_category',
    'Product Name': 'product_name',
    'Sales': 'sales',
    'Quantity': 'quantity',
    'Discount': 'discount',
    'Profit': 'profit'
}

df.rename(columns=new_column_names, inplace=True)
```

# **2. Thực hành query**

```python
# Câu 1: Hiển thị 5 dòng dữ liệu đầu tiên để xem cấu trúc bảng.
df.head()
```

```python
# Câu 2: Hiển thị xem thử 5 đơn hàng ở khu vực miền Tây, và chỉ cần biết thông tin về Vùng, Tên sản phẩm và Doanh số của các đơn đó.
result = df.loc[df['region'] == 'West', ['region', 'product_name', 'sales']].head()
print(result)
```

```python
# Câu 3: Tính Tổng doanh thu, Tổng số đơn hàng (mã order_id duy nhất) và Tổng số lượng khách hàng duy nhất của toàn bộ cửa hàng.
pd.DataFrame({
    'total_sales': [df['sales'].sum()],
    'total_orders': [df['order_id'].nunique()],
    'unique_customers': [df['customer_id'].nunique()]
})
```

```python
# Câu 4: Phân tích Vùng (region) nào mang lại doanh số lớn nhất, sắp xếp từ cao xuống thấp.
(df.groupby('region')['sales']
   .sum()
   .reset_index(name='total_sales')
   .sort_values(by='total_sales', ascending=False)
)
```

```python
# Câu 5: Xác định Danh mục (category) nào tạo ra nhiều lợi nhuận nhất.
(df.groupby('category')['profit']
   .sum()
   .reset_index(name='total_profit')
   .sort_values(by='total_profit', ascending=False)
)
```

```python
# Câu 6: Tìm 5 sản phẩm (product_name) có tổng doanh số cao nhất.
(df.groupby('product_name')['sales']
   .sum()
   .reset_index(name='total_sales')
   .sort_values(by='total_sales', ascending=False)
   .head(5)
)
```

```python
# Câu 7: Gom nhóm kết hợp theo category và sub_category để xem chi tiết mảng nào mang lại doanh thu cao nhất.
(df.groupby(['category', 'sub_category'])['sales']
   .sum()
   .reset_index(name='revenue')
   .sort_values(by='revenue', ascending=False)
)
```

```python
# Câu 8: Trích xuất năm từ cột ngày đặt hàng (order_date) và tính tổng doanh thu theo từng năm để xem đà tăng trưởng.
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')

(df.groupby(df['order_date'].dt.year)['sales']
   .sum()
   .reset_index(name='revenue')
)
```

```python
# Câu 9: Tính tổng doanh thu chi tiết theo từng tháng, nhưng chỉ giới hạn dữ liệu trong năm 2018.
# Tạo cột order_month ở định dạng YYYY-MM
# 1. Đảm bảo cột định dạng datetime
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')

# 2. Lọc, Tạo cột tháng và Gom nhóm
result = (
    df[df['order_date'].dt.year == 2018]                    # Lọc riêng năm 2018
    .assign(order_month=df['order_date'].dt.to_period('M')) # Tạo cột tháng-năm
    .groupby('order_month')['sales']                        # Gom nhóm theo tháng
    .sum()                                                  # Tính tổng doanh thu
    .reset_index(name='total_sales')                        # Đưa về dạng bảng
)

result
```

```python
# Câu 10: Phòng Marketing muốn gửi quà tri ân cho các khách hàng VIP.
# Cần tìm danh sách các khách hàng (customer_name) có tổng doanh số (sales) mua hàng từ trước đến nay lớn hơn 14400, và sắp xếp danh sách này từ cao xuống thấp.
result = (
    df.groupby('customer_name')['sales']
    .sum()
    .reset_index(name='revenue')
    .loc[lambda x: x['revenue'] > 14400]
    .sort_values(by='revenue', ascending=False)
)

result
```
